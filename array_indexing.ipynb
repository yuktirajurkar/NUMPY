```python
import numpy as np
```

NUMPY INDEXING AND SLICING OF VECTORS

```python
arr=np.arange(11,21)
arr
```

```python
arr[9]  #This will print the element at 9th index. This is INDEXING.
```

```python
# SLICING is take out a portion from an array.
arr[1:5] #This will give elemts from index 1 to n-1 that is 4 (5-1) in this case.
```

```python
arr[:5] # Starting index is 0 by default.
```

```python
arr[3:] # end is by default the last index of the array.
```

```python
arr_slice=arr[1:8:2] # this slices elements form index 1 to 8 at a gap of 1, say its taking jump of one.
arr_slice # sliced array can nbe stored in diff vatiable and perform further operations.
```

NUMPY INDEXING AND SLICING OF MATRIX

```python
arr=np.arange(1,31).reshape(6,5)
arr
```

```python
arr[5]  # to print a row
```

```python
arr[0,0] # to print a particular element.
```

```python
arr[0:2,1:3]  # to slice out some portion, 0:2 for row from 0 to 1 and 1:3 is for the column from 1 to 2. 
```

```python
#Similarly
arr[3:,3:]
```

```python
#to slice out the column, entire row has to be selected and desired column.
arr[:,2]
#It gives 
```

BOOLEAN INDEXING

```python
arr=np.arange(11,21)
arr
```

```python
bool_index=arr%2==0  
bool_index  #This will show true in place of the values that are divisible by 2 and falso otherwise.
```

```python
arr=arr[bool_index] # This will store the array having only the true values in arr. Al the false values are removed
arr
```
