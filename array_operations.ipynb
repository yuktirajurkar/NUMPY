DEEP AND SHALLOW COPY

```python
a =np.arange(1,21)
a
```

```python
slice = a[:5] # sliced out a portion from the array a.
slice = slice*10 # Did broadcasting in the sliced portion.
slice # printed the sliced array to see the result.
```

```python
a # But even after slicing the original array remains unaffected, it is still the same.
```

```python
# SHALLOW copy is when even after making changes to the sliced portion the orriginal array remains same because it creates a different copy or has a different space
# for sliced portion in RAM/Memory unlike list where the original list used to get affected.
```

```python
a[4:6]*10
```

```python
a  #Still the same in orginal array.
```

```python
a
b=a
```

```python
b[0]=99
b
```

```python
a # now a original array is changed because we changed b. Because when we say a=b they share the same location in memory. Therefore changes made in a are reflected in b.
  # This is called as DEEP copy.
```

```python
# if we do not want changes made in b to get reflected in a and want a and b to be the same, we use its copy.
b=a.copy()
b[2]=-25
b  # index 2 of b is changes to -25.
```

```python
a # remains same, therefore a shallow copy.
```

ADVANCE ARRAY MANIPULATIONS

1. Stacking array

```python
# matrix cannot be stacked therefore we stack two vectors.
a=np.array([1,2,3,4])
b=np.array([5,6,7,8])
```

```python
np.vstack((a,b)) #.vstack to stack vertically in 2-d.
```

```python
np.hstack((a,b)) # stacks array horizontally in 1d.
```

```python
np.column_stack((a,b)) # stacks arrays column-wise in 2d.
```

SPLITTIGN ARRAY

```python
c=np.arange(16).reshape(4,4)
c
```

```python
np.hsplit(c,2) # it divides array horizontally in two equal parts. Similarly it can be splitted in 4 equal parts, but 3 will give error.
```

```python
np.vsplit(c,4) # Splits array in 4 equal parts.
```
