```python
b=np.arange(1,11)
b #will create array from 1 to n-1 that is in this case 10 (11-1).
```

```python
b=np.zeros(6) #creates a vector with 6 zeroes.
b
```

```python
np.random.randint(6) # Generates any integer randomly between 0 to 6.
np.random.randint(10,20) # Generates any random int between 10 and 30.
np.random.randint(10,20,7) #Generate 7 random integers between 10 and 20.
```
